**Small example on KNN-mathmatics, iris-dataset**

```python
# usual imports
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# load the iris dataset
df = sns.load_dataset("iris")
```

```python
df
```

```python
# 50 of each flower type: setosa, versicolor and virginica
df['species'].value_counts()
```

```python
# we are going to only use sepal width and sepal length for our math demonstration
df = df[['sepal_length', 'sepal_width', 'species']]
```

```python
df
```

```python
# new data from user (or something similar)
sl_new = 7.1
sw_new = 3.6

# so what is the species for this flower with these measurements?
# => most probable species based on these values
```

```python
# calculate the Euclidean distance with the current dataset and new values
# above, original formula is in the materials
df['distance'] = np.sqrt((sl_new - df['sepal_length']) ** 2 + (sw_new - df['sepal_width']) ** 2)
```

```python
df
```

```python
# sort the data based on distance
# based on distance, assign each training data flower a rank
# closest neighbor => 1, furthest neighbor => highest rank
df = df.sort_values("distance", ascending=True)
df = df.reset_index()
df = df.drop("index", axis=1)
df['rank'] = df.index + 1
df
```

```python
# k-value => how many nearest neighboring rows 
# we want to take into account
# when we make our decision of the output
k_value = 10

# just show the relevant rows based on our k-value
sample = df[df['rank'] <= k_value]
sample
```
